## 課題1: 多層モデルの設計

重み付きReLUネットワーク`ReLUWeightedSum`は以下のスカラー関数を実現した．

$$f(x)=\sum_{i=1}^{K} w_i\,\operatorname{ReLU}(x-a_i)$$

これは1変数入力・1変数出力，つまり，$f: \mathbb{R} \to \mathbb{R}$という関数である．これを次のように一般化したい．

1. **m変数入力・n変数出力にしたい．** つまり，$F: \mathbb{R}^m \to \mathbb{R}^n$にしたい．

2. **多層化したい．** つまり，上記の$F$を繋げて，$\mathcal{F} = F_{\text{out}} \circ F \circ \cdots \circ F_{\text{in}}$のようにしたい．

> 項目1の多変数モデルを`ReLUWeightedSumLayer`というクラスで実装せよ．項目2の多層モデルを`ReLUWeightedSumNetwork`というクラスで実装せよ．


**備考**
- 記号$`\circ`$は関数の合成の意味で，$(f\circ g)(x)$は，要は$f(g(x))$という意味．
- 設計として，以下のように設計すれば，全体として$\mathcal{F}: \mathbb{R}^m \to \mathbb{R}^n$となる
    - $F_{\text{in}}: \mathbb{R}^m \to \mathbb{R}^d$
    - $F: \mathbb{R}^d \to \mathbb{R}^d$
    - $F_{\text{out}}: \mathbb{R}^d \to \mathbb{R}^n$

## 調査のヒント

学習の過程を観察する際の参考資料です．各文献の概要と，自分の実験で確かめられそうな点を挙げます．

**万能近似に関して**

- [ニューラルネットワークと深層学習 第4章：ニューラルネットワークが任意の関数を計算できることの視覚的証明](https://nnadl-ja.github.io/nnadl_site_ja/chap4.html)（Michael Nielsen，日本語訳）  
  小さな部品となる関数を組み合わせ，複雑な関数を近似する考え方を図と操作できるデモで説明しています．授業のReLUによる関数の構成の元にした解説です．「表現できること」と「学習でその関数に到達できること」を分けて考える参考になります．

**低周波数から学ぶ傾向**

- [On the Spectral Bias of Neural Networks](https://proceedings.mlr.press/v97/rahaman19a.html)（Rahamanら，ICML 2019）  
  ニューラルネットワークが，細かく変動する高周波成分より，なだらかに変動する低周波成分を先に学ぶ傾向を報告しています．異なる周波数の正弦波を足した目標関数を用意し，どの成分から近似が進むかを観察する参考になります．
- [Frequency Principle: Fourier Analysis Sheds Light on Deep Neural Networks](https://arxiv.org/abs/1901.06523)（Xuら，2019年初版）  
  学習が低周波から高周波へ進む傾向を「Frequency Principle」として調べています．学習途中の予測をフーリエ解析し，周波数ごとの誤差を比較することで，アニメーションで見えた変化を数値でも調べられます．

**簡単なパターン・サンプルから学ぶ傾向**

- [A Closer Look at Memorization in Deep Networks](https://proceedings.mlr.press/v70/arpit17a.html)（Arpitら，ICML 2017）  
  深層ネットワークはランダムなデータも記憶できますが，実データでは共通する単純なパターンを優先して学ぶ傾向を報告しています．また，実データには学習初期から正解しやすいサンプルと難しいサンプルがあることを調べています．回帰の課題へ応用するなら，滑らかな部分と複雑な部分，ノイズの有無などで誤差の減り方を比較できます．

**ネットワークの幅と学習ダイナミクス**

- [Neural Tangent Kernel（NTK）の概要](https://medium.com/lsc-psd/neural-tangent-kernel-ntk-%E3%81%AE%E6%A6%82%E8%A6%81-faf0ad249923)（Kai，2020，日本語解説）  
  ネットワークの幅を変えたときの学習の様子を，図やアニメーションで説明しています．記事の実験では，幅が大きいほど重みの初期値からの相対的な変化が小さくなる様子を観察できます．自分のモデルでも幅を変え，損失・予測関数・重みの変化を比較する参考になります．
