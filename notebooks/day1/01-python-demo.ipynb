### 2. リストと添字
添字は0から始まります。`a[start:stop]` は `stop` を含みません。

```python
scores = [60, 80, 50, 90]
print(scores[0])
print(scores[-1])
print(scores[1:3])
scores[2] = 70
print(scores)
```

### 3. 条件分岐とループ
`if` は条件に応じて処理を分けます。`for` はリストの各要素を順に処理します。インデントが処理の範囲を決めます。

```python
passed = 0
for score in scores:
    if score >= 70:
        passed += 1
print("70点以上:", passed)
```

### 4. 関数
関数は入力を受け取り、処理結果を返します。

```python
def relu_scalar(x):
    if x > 0:
        return x
    return 0

for value in [-2, 0, 3]:
    print(value, "->", relu_scalar(value))
```

### 6. クラス
クラスは、値とその値を使う操作をまとめます。`self` は作成したオブジェクト自身を表します。

```python
class LinearRule:
    def __init__(self, weight, bias):
        self.weight = weight
        self.bias = bias

    def predict(self, x):
        return self.weight * x + self.bias

rule = LinearRule(weight=2, bias=1)
print(rule.predict(3))
rule.weight = -1
print(rule.predict(3))
```

### 6. エラーを修正する
次のセルを実行するとエラーが出ます。原因を直して、`[80, 90]` が表示されるようにしてください。


```python
scores = [60, 80, 50, 90]
settings = {"bonus": 0, "threshold": 70}

def add_bonus(score, bonus):
    return score + bonus

passed = []
for score in scores:
    adjusted = add_bonus(score, settings["bonus"])
    if adjusted >= settings["threshold"]:
        passed.append(adjusted)
      print(passed)
```
