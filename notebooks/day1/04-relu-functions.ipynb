## 1. ReLUの定義
`ReLU`（Rectified Linear Unit）は、入力が負のときは0、正のときは入力をそのまま返す関数です。

$\operatorname{ReLU}(x)=\max(0,x)$

`ReLU(x-a)`では、折れ曲がる位置が`x=a`になります。スライダーを動かして確認してください。


```python
show_relu_single()
```

## 2. 3個のReLUで三角形の山を作る
3つのReLUを

$\operatorname{ReLU}(x-a) - 2\operatorname{ReLU}(x-b) + \operatorname{ReLU}(x-c)$

の形で足し合わせると、三角形の山を作れます。`a`、`b`、`c`を動かして、3つの項と合計がどう変わるか確認してください。


```python
show_relu_triangle()
```

$$
