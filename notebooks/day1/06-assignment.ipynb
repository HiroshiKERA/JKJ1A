## 課題1: 多層モデルの設計

重み付きReLUネットワーク`ReLUWeightedSum`は以下のスカラー関数を実現した．

$$f(x)=\sum_{i=1}^{K} w_i\,\operatorname{ReLU}(x-a_i)$$

これは1変数入力・1変数出力，つまり，$f: \mathbb{R} \to \mathbb{R}$という関数である．これを次のように一般化したい．

1. **m変数入力・n変数出力にしたい．** つまり，$F: \mathbb{R}^m \to \mathbb{R}^n$にしたい．

2. **多層化したい．** つまり，上記の$F$を繋げて，$\mathcal{F} = F_{\text{out}} \circ F \circ \cdots \circ F_{\text{in}}$のようにしたい．

> 項目1の多変数モデルを`ReLUWeightedSumLayer`というクラスで実装せよ．項目2の多層モデルを`ReLUWeightedSumNetwork`というクラスで実装せよ．


**備考**
- 記号$`\circ`$は関数の合成の意味で，$(f\circ g)(x)$は，要は$f(g(x))$という意味．
- 設計として，以下のように設計すれば，全体として$\mathcal{F}: \mathbb{R}^m \to \mathbb{R}^n$となる
    - $F_{\text{in}}: \mathbb{R}^m \to \mathbb{R}^d$
    - $F: \mathbb{R}^d \to \mathbb{R}^d$
    - $F_{\text{out}}: \mathbb{R}^d \to \mathbb{R}^n$
