# Day2 発展課題（最終レポートに含める）

> **AIモデルは画像をどの程度理解しているのだろうか？ また何を理解できていないだろうか？**

この疑問に基づいて，仮説・課題を設定し，調査・実験により検証せよ．

### **課題**

Day2では，人間であれば決して間違えないであろう設定において画像モデルが失敗する例をいくつか見ました．「高精度なAIモデル」が完全に信頼できるものではないということが実感できたと思います．

このような例をさらに発見し，そのパターンなどを分析して下さい．以下に実験に有効な情報をまとめます

---


#### **モデル・データのヒント**
- [Hugging Face](https://huggingface.co/)にはさまざまな学習済みモデルとデータが公開されています．基本的に，どのモデルも数行のコードで簡単に利用でき，使用例が示されています．
- [HuggingFaceで簡単AI入門](https://nonstop-umbrella-d21.notion.site/HuggingFace-AI-974bca6b55ae4991a97010e4e3dd24ff)．簡単な概要に関しては，旧講義ページにもまとめがあります．

#### **調査のヒント**

さまざまな論文やブログ記事で，モデルの失敗例が報告されています．以下に参考となる例をいくつか並べています

- [Explaining and Harnessing Adversarial Examples](https://arxiv.org/abs/1412.6572)（Goodfellowら，2014／2015）  
  画像に小さな変化を加え，分類モデルを誤らせる敵対的入力を扱う論文です．授業で使ったFGSMを提案し，入力に関する勾配から攻撃を作る方法を示しています．
- [Adversarial examples in the physical world](https://arxiv.org/abs/1607.02533)（Kurakinら，2016／2017）  
  敵対的画像をカメラで撮影して入力しても，誤分類が起こることを調べた論文です．I-FGSMの背景資料でもあり，画像を直接入力する場合と，撮影を介する場合の違いを考える参考になります．
- [Are DNNs fooled by extremely unrecognizable images?](https://arxiv.org/abs/2012.03843)（Kumano・Kera・Yamasaki，2020／2022）  
  単色画像の少数の画素を変えるだけで，人間には対象が分からない画像をモデルが高い確信度で分類する現象を調べています．自然画像への微小な変化とは異なる方法で，モデルの認識を探る例です．
- [Multimodal neurons in artificial neural networks](https://openai.com/index/multimodal-neurons/)（OpenAI，2021，研究紹介記事）  
  画像と文字を結び付けるCLIPの内部表現を紹介しています．画像内に別の物体名を書き込むと，見た目より文字に引きずられて分類する「typographic attack」の例があります．
- [Defense-Prefix for Preventing Typographic Attacks on CLIP](https://arxiv.org/abs/2304.04512)（Azuma・Matsui，2023）  
  画像中の文字によってCLIPを誤らせる攻撃への対策を扱います．モデル本体を変更せず，クラス名の前に学習したprefixを加える方法で，文字の影響を抑えることを目指しています．
- [Vision language models are blind](https://vlmsareblind.github.io/)（Rahmanzadehgerviら，2024）  
  線の交差，円の重なり，物体の数など，人間には単純に見える視覚課題でVLMが失敗することを調べています．図形の配置や数を変え，どのような条件で正しく答えられるかを比較する参考になります．
- [Vision Language Models are Biased](https://arxiv.org/abs/2505.23941)（Voら，2025）  
  動物の足の本数などを通常とは異なる状態にした画像で，VLMが実際の画像よりも既知の知識に沿って答えてしまう傾向を調べています．授業で使った鹿の画像の出典です．[公式ページ](https://vlmsarebiased.github.io/)にも例があります．
- [Unveiling Typographic Deceptions: Insights of the Typographic Vulnerability in Large Vision-Language Model](https://arxiv.org/abs/2402.19150)（Chengら，2024）  
  画像に書かれた文字によって，文章で回答するVLMが誤った答えへ誘導される現象を調べています．文字の大きさ・色・位置などを変え，攻撃の効き方を比較する実験の参考になります．
- [SceneTAP: Scene-Coherent Typographic Adversarial Planner against Vision-Language Models in Real-World Environments](https://openaccess.thecvf.com/content/CVPR2025/html/Cao_SceneTAP_Scene-Coherent_Typographic_Adversarial_Planner_against_Vision-Language_Models_in_Real-World_CVPR_2025_paper.html)（Caoら，2025）  
  場面に自然に見える文字を組み込み，VLMの回答を誘導する攻撃を提案しています．画像と質問に応じて攻撃文や配置を考える方法で，単に物体名を貼り付ける攻撃との違いに注目できます．
- [Universal and Transferable Adversarial Attacks on Aligned Language Models](https://llm-attacks.org/)（Zouら，2023）  
  質問の末尾に付ける文字列を自動的に探索し，LLMが通常は拒否する要求にも回答してしまう現象を調べています．あるモデルで作った攻撃が別のモデルにも効く「転移性」も検証しており，画像への敵対的入力と文章への攻撃を比較する参考になります．



**論文を読むにあたり**

文献調査にあたってはAIを使いうまく読んでください．技術的な部分は読むのが大変ですので，そのあたりは気にせず以下の点に集中して情報を取るようにして下さい．
- **課題設定**：AIモデルのどのような振る舞いに着目しているのか？
- **動機**：その振る舞いにはなぜ着目する価値があるのか？
- **アイディア**：この研究（特に手法や実験設定）の技術的なアイディアはなんなのか？
- **洞察**：この研究が与えている洞察はなんなのか（Takeaway messageは何か）

この情報は通常，論文のAbstract / Introduction / Conclusion にまとめられています．

***注：** AIに普通に論文を渡して聞くと，かなり高度な説明がされると思います．うまく説明の難易度を調整するプロンプトを作りましょう
